```python
import torch
import torch.nn.functional as F
import xarray as xr

data = xr.open_zarr("../data/processed/antarctic/all/all.zarr")

true = torch.tensor(data.d_labels.values)
persistence = torch.zeros_like(true)

mse = F.mse_loss(persistence, true)
mae = F.l1_loss(persistence, true)
rmse = torch.sqrt(mse)

print(f"MSE: {mse * 100} cm/s\nMAE: {mae * 100} cm/s\nRMSE: {rmse * 100} cm/s")
```
